```python
from dataclasses import dataclass, field
```

```python
@dataclass
class User:
    features: list = field(default_factory=list)
```

```python
u1 = User()
```

```python
u1.features.append(22)
```

```python
u2 = User()
```

```python
print(u2.features)
```

```python
print(u1.features)
```
